# 넘파이를 사용하는 이유
* 행렬 연산이 쉽고 빠르기 때문

```python
# 스칼라 연산
a = 10
b = 5
scalar = a + b
print(scalar)
```

```python
scalar + 10
```

```python
# 1차원 데이터 벡터의 연산
b1 = [1,2,3,4,5]
# 리스의 요소에 10을 곱한 후 다시 리스트로 만드세요.
result = []
for num in b1:
    result.append(num*10)
result
```

```python
# 리스트 표현식으로도 계산은 가능
[num * 10 for num in b1]
```

```python
# 안에 있는 요소에 곱셈을 하려면 반복문으로만 가능하다.
b1 * 10
```

```python
# 2차원 데이터 계산 * 10하기
second = np.array([[1,2,3],[1,2,3]])
print(second, second.ndim, second.shape)
```

```python
result = []
for num in second:
    mid_result = []
    for number in num:
        mid_result.append(number * 10)
    result.append(mid_result)
result = np.array(result)
```

```python
result.ndim
```

```python
# 3차원 데이터의 요소에 10 곱하기
third = np.array([[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]])
print(third, third.ndim, third.shape)
```

```python
result = []
for num1 in third:
    num1_result = []
    for num2 in num1:
#         print(num2)
        num2_result = []
        for num3 in num2:
            num2_result.append(num3 * 10)
        num1_result.append(num2_result)
    result.append(num1_result)
result = np.array(result)
```

```python
result
```

```python
result.ndim
```

```python
# 4차원 데이터 요소에 10 곱하기
forth = np.array([[[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]],
                 [[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]]])
print(forth.ndim, forth.shape)
```

```python
forth * 20
```

```python
# 5차원 데이터 요소에 10 곱하기
fifth = np.array([[[[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]],
                 [[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]]],
                 [[[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]],
                 [[[1,2,3], [1,2,3]],[[1,2,3], [1,2,3]]]]])
print(fifth.ndim, fifth.shape)
```

* argsort() 자료 중 값이 작은 것에서 큰 순으로 원래 인덱스 출력<br>
**원자료의 손상없이 정렬할 수 있다**

```python
g = np.array([5, 0.5, -7, 42, 10, 0])
print(g)
```

```python
# 2, 5, 1, 0, 4, 3 (0.5[2] 부터 10[5]까지)
```

```python
g.argsort()
```

```python
s = g.argsort()
s
```

```python
g[s][::-1]
```
