* for 반복문을 이용해서 1부터 10까지 정수를 더한 합을 출력하세요
  * 1부터 10까지 숫자를 만들어서 반복문에 넣는다.
  * 1부터 10까지 누적 덧셈을 한다.
  * 누적 덧셈을 한 결과를 출력한다.

```python
hap = 0
for num in range(1, 11):
#    print(num)
    hap = hap + num
#for문과 같은 열로 이동시키기(반복문 블럭에서 벗어나기)
print(hap)
```

* 반복문은 열의 위치로 블록을 구분할 수 있다.

```python
# hap을 반복문 바깥으로 빼야 영향을 받지 않는다.
hap = 0
for num in range(1, 11):
#    hap=0
#    print(num, end=" ")
    print(f"hap:{hap} + num:{num}", end=": ")
    hap += num # hap = hap + num
    print(f"{hap}")
```

* 1부터 50까지 수 중에서 짝수인 수의 합을 구하시오
* 1-50 수를 만들기 (range(1-51))
* 홀수, 짝수 구분 if, % 2 == 0
* 짝수일 때 누적 합, 결과저장할 변수 hap
* 최종 결과 출력

```python
hap = 0
for num in range(1, 51):
#print(num, end=" ")
    if num %2 == 0:
        print(f"hap:{hap} + num:{num}", end=": ")
        hap += num # hap = hap + num
        print(f"{hap}")
```

```python
hap = 0
for num in range(1, 51):
#    print(num, end=" ")
    if num % 2 == 0:
        print(f'hap:{hap}, num:{num}')
        hap += num
print(hap)
```

* 1부터 100까지 수를 만들고 그 중에서 30-70까지만 더해서 출력하세요잇
* 과정 출력
* 2가지 방법을 이용해서 코드 작성 if 조건문, 슬라이싱

```python
# if 조건문
hap = 0
for num in range (1, 101):
#     print(num, end=" ")
    if 30 <= num <=70:
        print(f'hap:{hap}, num{num}')
        hap += num
print(hap)
```

```python
# 슬라이싱
hap=0
for num in range(1,100)[29:70]:
#     print(num, end=" ")
    print(f'hap{hap}, num{num}')
    hap += num
print(hap)
```

```python
# 슬라이싱 (step으로 홀짝 구분 가능)
hap=0
for num in range(1,100)[29:70:3]:
#     print(num, end=" ")
    print(f'hap{hap}, num{num}')
    hap += num
print(hap)
```

## continue, break
* continue: 코드 실행을 건너뛰고 반복 계속 진행
* break: 특정 조건이 되었을 때 반복문 정지


```python
for idx, i in enumerate(s):
#     print(f"현재 반복 횟수: {idx+1}/{len(s)}", end = "\r")
    if idx %2 == 0:
        continue
    else:
        print(idx, i)
        if idx >= 19:
            break

    time.sleep(1)
#     print(idx, i)
```

## while의 무한 반복

* while 뒤의 조건이 True인 경우 무한 반복
* while문 내에서 if 조건문으로 break를 걸어주어야 멈춤

* 무한 반복 후 while문 내에서 조건을 주어 멈추기

```python
# 무한 반복
while True:
    print("무한반복", end=" ")
```

```python
i = 1
while True:
    print(i, "조건 반복", end = " ")
    if i >= 100:
        break
    i += 1
```

* 무한 반복이 되는 조건: True
* True가 되는 경우: 1, 정수, 실수, 문자열, list, tuple, dict, True
* False가 되는 경우: 0, 0.0, "", '', [], (), {}, False

```python
#보통은 True로 쓴다
i = 1
while " ":
    print(i, "조건 반복", end = " ")
    if i >= 100:
        break
    i += 1
```

# 중첩 반복문
* 반복문을 2개 이상 겹쳐서 사용
* 바늘 시계의 원리와 같음
* 안쪽 반복문이 끝난 후 바깥쪽 반복문이 실행
* 초침 -> 분침 -> 시침

* 구구단 만들기 2중 중첩 for문
* 2단 - 9단
* 1-9
* 2*1 = 2

```python
for num1 in range(2, 10):
#     print(num1)
    for num2 in range(1, 10):
        print(f"{num1} X {num2} = ", num1 * num2)
```

* 반복문 3개로 시계 만들기
* 초 00 - 59
* 분 00 - 59
* 시 00 - 23
* import time
* time.sleep(0.2)

```python
# 3중 반복문 하나로, 0.2초당 3600번씩 코드가 반복실행된다.
import time
for si in range(0, 24):
    for bun in range(0, 60):
        for cho in range(0, 60):
            time.sleep(0.1)
            print(f"{si:02d}:{bun:02d}:{cho:02d}", end="\r")      
```

```python
apart = [[101, 102], [201, 202], [301, 302]]
```

```python
# list에서 자료를 for문으로 1개씩 꺼내기
# 꺼낸 자료를 다시 for문에 넣어서 안쪽 자료 꺼내기
# 자료 + "호" 붙여서 출력하기
for num1 in apart:
#     print(i)
    for num2 in num1:
        print(f"{num2} 호")
```

아래와 같이 출력되게 수정해보기

```python
#슬라이싱으로 인덱스 역으로 주기
for num1 in apart[::-1]:
#     print(i)
    for num2 in num1:
        print(f"{num2} 호")
```

```python
# 층있는 바깥 리스트의 순서를 역순 정렬
# 반복문으로 하나씩 꺼내기
# 꺼낸 자료 다시 반복문에 넣고 출력하고 "호" 붙이기
for i in reversed(apart):
    for j in i:
        print(j, "호")
```
