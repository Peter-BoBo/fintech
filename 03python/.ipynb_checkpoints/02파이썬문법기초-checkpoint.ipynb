# print 문
* print 함수 안에 있는 내용을 출력해주는 함수
* print(출력할 내용)
* print(1) 출력할 내용이 숫자인 경우, 변수인 경우 따옴표 X
* print("hello world") 출력할 내용이 문자, 문자열인 경우 반드시 따옴표 필요. 단 "",'' 모두 사용가능

```python
print(1)
```

```python
print(3.14)
```

```python
print("파")
```

```python
print('파이썬')
```

```python
# 따옴표까지 같이 출력하고 싶을때는 섞어서
print("이것은 '파이썬'이다."")
```

```python
print('이것은 "파이썬"이다.')
```

# print문의 옵션
* sep: 구분자 - print문 안에 여러개의 자료를 출력할 때 자료를 구분해주는 문자
* 기본값은 ""공백

```python
print("이것은", '파이썬', '입니다.')
```

```python
print("이것은", '파이썬', '입니다.', sep="_")
```

```python
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

* end 옵션 - print가 끝나고 실행될 옵션
* 기본값 = \n - 이스케이프 문자 new line 새줄

```python
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

```python
# end 값을 공백 " " 으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end=" ")
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

```python
# end 값을 공백 "\t" (tap)으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end="\t")
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

```python
# end 값을 공백 "\r" (cariige return)으로 준 경우
# 마지막 줄만 출력됨 (반복문 시 유용)
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end="\r")
print('스파이더맨', "헐크", '아이언맨', sep=" vs ")
```

```python
# end 값을 공백 "\n" (new line)으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end="\n\n\n")
print('스파이더맨', "헐크", '아이언맨', sep=" vs ")
```

```python
# 문자열과 숫자
print("헐크", 1, 3.14, sep=" - ", end="\n\n\n")
print('스파이더맨', "헐크", '아이언맨', sep=" vs ")
```

# 변수
* 자료를 담는 택배상자
* 택배 상자는 1개만 넣을 수 있다.
* 변수는 영문과 숫자를 조합해서 만들 수 있다.
* 변수명의 시작은 반드시 영문자로 시작 or _로 시작
* 변수명이 숫자로 시작하거나 -등의 특수문자로 시작하면 오류
* 예약어(이미 파이썬에서 사용하고 있는 함수명, 문법용 단어 ex: for, list,if, id)는 변수명으로 사용할 수 없다.
    * 주의!! 예약어를 써도 파이썬에서는 변수가 만들어진다. 
    * 그러므로 예약어를 잘 기억해두자!!

```python
a = 1
```

```python
id(a)
```

```python
print(id)
```
