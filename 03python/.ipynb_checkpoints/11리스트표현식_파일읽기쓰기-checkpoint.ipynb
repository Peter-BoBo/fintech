# 리스트 표현식
* 리스트 안에 for 반복문과 if else 조건문을 한 번에 적어서 리스트를 만드는 방법
* 여러 줄의 코드를 1줄로 줄일 수 있다.
* for문과 if else문을 사용해서 처리할 때 보다 속도가 빠르다.
* [실행할 명령 for 변수 in list, tuple]

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(f"{i}는 짝수입니다.")
    else:
        print(f"{i}는 홀수입니다.")
```

* 1부터 10까지 들어있는 리스트 만들기

```python
l1 = []
for i in range(1, 11):
    l1.append(i)
l1
```

```python
[i for i in range(1, 11)]
```

## 리스트 표현식과 일반 반복문의 속도 차이

```python
import time
```

```python
start = time.time()
a = []
for i in range(1, 200000000):
    a.append(i + 5) 
print(a[:10])

print("코드 실행 시간: ", time.time() - start)
```

```python
%%time
# start = time.time()
a = []
for i in range(1, 200000000):
    a.append(i + 5) 
print(a[:10])

print("코드 실행 시간: ", time.time() - start)
```

```python
start2 = time.time()
b = [i+5 for i in range(1, 20000000)]
print(b[:10])
print("코드 실행 시간: ", time.time() - start2)
```

```python
%%time
# start2 = time.time()
b = [i+5 for i in range(1, 20000000)]
print(b[:10])
# print("코드 실행 시간: ", time.time() - start)
```

```python
import numpy as np
```

```python
start3 = time.time()

c = np.arange(1, 200000000)
c = c + 5
print(c[:10])
print("코드 실행 시간: ", time.time() - start3)
```

```python
# start3 = time.time()

c = np.arange(1, 200000000)
c = c + 5
print(c[:10])
# print("코드 실행 시간: ", time.time() - start3)
```

* 1에서 10까지의 숫자중 짝수만 더해서 리스트 만들기

```python
l1 = []
for i in range(1, 11):
    if i % 2 == 0:
        l1.append(i)
l1
```

```python
l2 = [i for i in range(1, 11) if i % 2 == 0]
l2
```

# 파일 읽기 쓰기 open(), with open()
## 1) 파일에 문자열을 쓰거나 읽을 때는 open 함수 사용
* 파일저장
변수명 = open("파일이름", 모드(w:쓰기, r:읽기)<br>
변수명.write("저장할 내용")<br>
변수명.close()<br>
**close()를 안하면 계속 열려있음**

```python
file = open("test.txt", "w")
file.write("파일 만들기 연습")
file.close()
```

* 파일열기
변수1 = open("파일이름", 파일모드(r))<br>
변수2 = 변수1.read()<br>
변수1.close()

```python
file1 = open("test.txt", "r")
my_file = file1.read()
file1.close()
print(my_file)
```

## with open
* with open은 close()를 하지 않아도 자동으로 close()처리를 해줌

```python
with open("test2.txt", "w")as file3:
    file3.write("with open을 사용하면 파일을 자동으로 닫아주어 편리하다.")
```

```python
with open("test2.txt", "r")as file4:
    data = file4.read()
print(data)
```

# encoding에 따른 읽기 오류
* 운영체제나(windows, linux, macOS), 한글/영어 윈도에 따라서 인코딩 되는 포멧이 달라짐
* 인코딩에 따른 오류가 발생 가능
* web표준은 utf-8이므로 인코딩은 utf-8로 하는 것이 좋다.
* windows의 경우 인코딩이 cp949이므로 주의

```python
with open("test_utf-8.txt", 'w', encoding='utf-8') as file5:
    file5.write("이것은 with open으로 만든 utf-8 인코딩 파일입니다. ")
```

```python
with open("test_utf-8.txt", 'r') as file6:
    file6_data = file6.read()
print(file6_data)
```
