```python
result3 = {}
for info in data[0]['body']['items']:
#     print(info)
    for key, value in info.items():
#         print(f"{key} : {value}")
        result3.setdefault(key, []).append(value)
#         print(result3)
```

```python
print(data[0]['header']['columns'],end="")
```

```python
# key 이름을 header - column 한국어 이름으로 변경하기

result4 = {}
for page in data: # 전체 데이터 페이지들을 순회 (37페이지를 1페이지 씩 차례로)
    for info in data[0]['body']['items']: # 첫 번째 페이지의 아이템들 순회
        for key, value in zip(page['header']['columns'], info.values()): # page의 헤더 컬럼과 item의 값을 동시에 묶어서 처리
            result4.setdefault(key, []).append(value)
#             print(key, value)
df2 = pd.DataFrame(result4)
df2
```
