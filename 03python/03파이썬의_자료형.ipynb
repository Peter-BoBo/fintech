# 파이썬의 자료형(data type) 
###  ※매우 중요※
## 파이썬의 자료형은 동적으로 변화 
### ※type()을 잘 확인하는 것이 중요하다※
* int: 정수(소숫점이 없는 숫자)
* float: 실수(소숫점이 있는 숫자)
* str: 문자/문자열
* list: 여러개의 자료가 순서대로 저장된 자료형, 모든 자료형을 넣어 만들 수 있음
* tuple: list와 비슷하나 한 번 만들면 수정이 안됨, 모든 자료형을 넣어 만들 수 있음
* dictionary(dict): "key : value" 쌍으로 이루어진 자료형, 모든 자료형을 넣어 만들 수 있음

# 딕셔너리형(dict) JSON{"key" : "value"}
### ※매우 중요하다 공부잘해놓자 진짜로!※
* key와 value로 이루어진 데이터형
* json 이라고 불리며 웹상에서 데이터를 주고 받을 때 주로 사용
* 딕셔너리의 value에는 모든 자료형이 올 수 있음
* 딕셔너리는 순서가 없는 자료형
* key를 호출하면 value가 출력됨 (인덱스는 번호! 딕셔너리는 키!)
* 데이터프레임을 만들때, 딕셔너리를 사용해 만든다. (웹크로울링시 사용)
* 딕셔너리를 잘 공부해놓자!


```python
dic1 = {'이름': "홍길동", "나이" : 33, "키" :  175.4, "가족" : ["둘리", "또치", "도우너"]}
dic1
```

```python
print(dic1)
```

```python
dic1['가족']
```

# 형변환 함수
* 형변환: data type 다른 data type으로 바꾸는 것
* 변수 자체의 값을 바꾸는 것은 아니다.
* 재할당을 하지 않는 이상, 변수 고유값은 동일하다.

```python
a = 3
b = "3"
```

```python
a
```

```python
b
```

```python
print(a,b)
```

```python
a+b
```

*정수로 형변환 int()<br>
*계산할 때만 자료형이 바뀌는 것

```python
a + int(b)
```

* 실수로 변환 float()

```python
a + float(b)
```

* 문자열로 변환 str()

```python
str(a) + b
```

* 리스트 형변환 list()

```python
t = (1,2,3,4)
print(t, type(t))
```

```python
t2.append(5)
print(t2)
```
