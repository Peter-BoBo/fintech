* num_maker2 - 정방향, 역방향 리스트가 출력되는 함수

```python
def num_maker(start, end, step):
    result = list(range(start, end, step))
    result2 = []
    print(result)
    for i in result:
        result2.append(i*10)
    print(result2)
    return result, result2
```

```python
def num_maker2(start, end, step):
    list_result = list(range(start+3, end+3, step))
    reversed_list =  list_result[::-1]    
    print(list_result, reversed_list)
    return list_result, reversed_list
```

# 중간에 함수를 종료시키기 위한 용도의 return
* 반복문에서 break 처럼 활용 가능
* 함수 내에서 return이 실행되면 함수는 종료

* 1에서 100까지 숫자 중에서 짝수만 더해서 출력하시오.
* 단, 숫자가 50이상이 되면 멈추시오.
* 더해지는 과정도 출력

```python
num = list(range(1, 101))
hap = 0
for i in num:
    if i < 50: 
        if i % 2 == 0:
            hap += i
            print(hap)
    else:
        break
```

```python
def even(num):
#     print(num)
    hap = 0
    for i in num:
        if i < 50: 
            if i % 2 == 0:
                hap += i
                print(hap, end=" ")
        else:
            break
    return hap
  
```

```python
even(list(range(1, 101)))
```

* break없이 멈추기

```python
# break 없이 return으로 반복을 중단시킬 수 있다.

def even2(num):
#     print(num)
    hap = 0
    for i in num:
        if i < 50: 
            if i % 2 == 0:
                print(f'{i}+{hap}=', end=" ")
                hap += i
                print(f'{hap}', end=" ")
        else:
            return hap
```

* 점수를 입력 받아서 91-100이면 A학점, 81-90 이면 B학점, 71-80이면 C학점, 61-70 D학점 60이하는 F학점을 출력하는 student_grade 함수를 만드세요.
* 점수를 입력받음 input()
* student_grade 함수에 입력받은 값을 넣어줌
* student_grade 함수에서는 입력받은 값을 비교해서 학점 출력

```python
def student_grade(score):
    score = int(input())
    if 91 <= score <= 100:
        print("A학점")
    elif 81 <= score < 91:
        print("B학점")
    elif 71 <= score < 81:
        print("C학점")
    elif 61 <= score < 71:
        print("D학점")
    elif score < 61:
        print("F학점")
    else:
        print("집중해라")
        
    return score
```

* 위치인수와 가변인수 동시에 받기
* 위치인수와 가변인수를 동시에 받을 때는 반드시 위치인수가 먼저 와야 한다.
* def muls(a, *args): OK
* def muls(*args, a): X

```python
# 정수와 리스트를 입력받아 리스트의 요소에 입력받은 정수를 곱하는 함수
```

```python
def muls(a, *args):
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
l2 = [1, 2, 3, 4, 5]
```

```python
muls(6, *l2)
```

```python
muls(*l2, 2)
```

```python
def muls(*args, a):
    print(a, args)
    result = []
    for num in args:
        result.append(a*num)
    return result
```

```python
muls(1,2,3)
```

```python
muls(*l2, 3)
```

* 1에서 10까지 수 중에서 3의 배수만 문자열로 바꾸기

```python
def tostr(x):
    result=[]
    for i in x:
        if i % 3 == 0:
            result.append(str(i))
        else:
            result.append(i)
    return result
```

```python
result_str = tostr(list(range(1, 11)))
print(result_str)
```
